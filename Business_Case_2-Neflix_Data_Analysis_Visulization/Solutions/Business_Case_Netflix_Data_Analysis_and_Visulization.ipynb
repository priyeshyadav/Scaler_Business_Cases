Now there are 2 things that we need to perform.<br>
1. Convert that data type to date, currently its Object. On performing this operation NaN value will change to NaT (Not a Time).<br>
2. New Nat value should be updated to most used dates, We are avoiding to use 'Unknown Date' or 0 as we are converting this column to datetime from object. Another option is to convert this NaT to a very old date but that could become outliers.

Lets check the format of the Date so that we can put it as a keyword argument.

```python
netflix_data['date_added'].head()
```

```python
netflix_data['date_added'] = netflix_data['date_added'].str.strip()
#We need to strip leading & trailing spaces as we are getting error in date format while converting it.
```

```python
netflix_data['date_added'] = pd.to_datetime(netflix_data['date_added'], format='%B %d, %Y')
```

```python
netflix_data['date_added'] = netflix_data['date_added'].fillna(netflix_data['date_added'].mode()[0])
#mode() might result in a series if there are ties, hence we are using the most frequent date
```

```python
print(f"Not missing values are : ",netflix_data["date_added"].notna().sum())
```
