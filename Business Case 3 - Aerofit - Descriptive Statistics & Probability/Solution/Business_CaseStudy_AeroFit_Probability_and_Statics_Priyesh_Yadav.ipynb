Lets draw the box plot to find outliers in our numarical columns. There are 6 numarical columns, lets plot them in 6 subplots.

```python
fig, axes = plt.subplots(2, 3, figsize=(10, 6))

# Manually assign each boxplot
sns.boxplot(y=row_data['Age'], ax=axes[0, 0])
axes[0, 0].set_title('Age')

sns.boxplot(y=row_data['Education'], ax=axes[0, 1])
axes[0, 1].set_title('Education')

sns.boxplot(y=row_data['Usage'], ax=axes[0, 2])
axes[0, 2].set_title('Usage')

sns.boxplot(y=row_data['Fitness'], ax=axes[1, 0])
axes[1, 0].set_title('Fitness')

sns.boxplot(y=row_data['Income'], ax=axes[1, 1])
axes[1, 1].set_title('Income')

sns.boxplot(y=row_data['Miles'], ax=axes[1, 2])
axes[1, 2].set_title('Miles')

plt.tight_layout()
plt.show()
```
